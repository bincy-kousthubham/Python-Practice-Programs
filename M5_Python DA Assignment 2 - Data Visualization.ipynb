### 2) Handling Missing Values
- Check for missing values in the dataset and identify columns with missing data.
- Impute missing values using appropriate strategies based on the column type (e.g., using mean, median, or mode for numerical columns, and mode for     categorical columns).
- For columns that are critical and cannot be reasonably imputed, remove rows with missing values to maintain data integrity.

```python
#Check for missing values in the dataset and identify columns with missing data
df.isnull().sum()
```

```python
#Impute missing values using appropriate strategies based on the column type (e.g., using mean, median, or mode for numerical columns, and mode for categorical columns).
print("Most frequent payment mode: ", df["payment"].mode()[0])
df["payment"]=df["payment"].fillna(df["payment"].mode()[0])
df.isnull().sum()
```

```python
missing_row_count = df.isnull().any(axis=1).sum()
print("Total number of Missing value rows: ", missing_row_count)
```

```python
#For columns that are critical and cannot be reasonably imputed, remove rows with missing values to maintain data integrity.
#Dropping the missing value rows
df1 = df.dropna(subset=['pickup_zone','dropoff_zone', 'pickup_borough', 'dropoff_borough'])
print("Number of rows and Columns after dropping missing rows: ", df1.shape)
df1.isnull().sum()
```

### 3) Visualizations using Matplotlib/Pandas Plot:

```python

#Line Chart Plot a line chart to visualize the fare over time, using the pickup timestamp as the x-axis and fare as the y-axis. Ensure the pickup column is converted to a datetime format before plotting.
import pandas as pd
import matplotlib.pyplot as plt

#checking pickup column for datetime format
print("Datatype of pickup column: ", df1["pickup"].dtype)
print(df1["pickup"].dtype=='datetime64[us]')

df1=df1.sort_values(by="pickup")
plt.figure(figsize=(12, 6))
plt.plot(df1["pickup"],df1["fare"])
plt.xlabel("Pickup Timestamp")
plt.ylabel("Fare")
plt.title("Taxi Fare over Time")
plt.xticks(rotation=60)
plt.grid(True)
plt.show()
```

```python
#Bar Chart - Create a bar chart to show the total fare for each pickup_borough. Group the data by pickup_borough and sum the fare for each group.
print("Datatype of pickup_borough column: ", df1["pickup_borough"].dtype)

borough_fare = df1.groupby("pickup_borough")["fare"].sum()
print(borough_fare)

plt.figure(figsize=(8,4))
plt.bar(borough_fare.index, borough_fare.values)
plt.xlabel("Pickup_borough")
plt.ylabel("Total Taxi Fare")
plt.title("Total fare per Pickup borough")

plt.xticks(rotation=60)
plt.grid(axis='y',linestyle='--', alpha=0.5)
plt.show()
```

```python
#Pie Chart - Plot a pie chart showing the distribution of trips based on the payment method (credit card, cash, etc.). 
#Each slice should represent the count of trips for a specific payment method.

payment_count = df1["payment"].value_counts()
print(payment_count)

plt.figure(figsize=(4,4))
plt.pie(payment_count.values,labels=payment_count.index, autopct="%1.1f%%", startangle=90)
plt.title("Trips distribution by payment method")
plt.show()
```

```python
#Histogram - Create a histogram to visualize the distribution of distance. 
#Customize the number of bins for better granularity and ensure the plot is easy to interpret.

print(df1["distance"].describe())
plt.figure(figsize=(10,6))
plt.hist(df1["distance"].dropna(), bins=20, edgecolor="black")
plt.xlabel("Trip Distance")
plt.ylabel("Frequency")
plt.title("Distribution of trip distances")
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()
```

```python
# Box Plot - Plot a box plot to visualize the distribution of tip amounts for each pickup_borough.
# Use pickup_borough as the categorical axis and tip as the numeric axis. 

print(df1["pickup_borough"].unique())
print(df1["tip"].isnull().sum())

plt.figure(figsize=(10,6))
df1.boxplot(column="tip",by="pickup_borough")
plt.xlabel("pickup_borough")
plt.ylabel("Tip Amount")
plt.title("Distribution of Tip amounts by pickup Borough")
plt.suptitle('')  # Remove automatic Pandas title
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### 4) Visualizations using Seaborn:

```python
# ★ Count Plot - Create a count plot to visualize the number of trips in each pickup_borough. 
# The x-axis should represent the boroughs, and the y-axis should show the count of trips.

plt.figure(figsize=(8, 4))
sns.countplot(data=df1, x="pickup_borough", hue="pickup_borough", legend=False, palette="dark")
plt.xlabel("Pickup Borough")
plt.ylabel("Number of Trips")
plt.title("Number of Taxi Trips by Pickup Borough")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python
# Scatter Plot - Plot a scatter plot to show the relationship between distance and fare. 
# Use distance on the x-axis and fare on the y-axis to visualize any correlation. 
# Color the points based on the pickup_borough to differentiate the trips by their respective boroughs.

sns.scatterplot(data=df1, x="distance", y="fare", hue="pickup_borough")
plt.title("Relationship between Distance and Fare")
plt.xlabel("Distance")
plt.ylabel("Fare")
plt.show()
```

```python
# Heatmap - Plot a heatmap to visualize the correlation between numerical variables such as distance, fare, tip, tolls, and total. 
# Use a correlation matrix to highlight the relationships.

corr_matrix = df1[['distance', 'fare', 'tip', 'tolls', 'total']].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Taxi Data')
plt.show()
```

```python
# Pair Plot - Create a pair plot to visualize the pairwise relationships between distance, fare, tip, and total. 
# Color the data points according to the pickup_zone method to compare how different zones affect these variables.

#As we have many pickup zones, creating the pair plot for selected 5 pickup zones.

selected_zones = df1['pickup_zone'].value_counts().head(5).index
df_sample = df1[df1['pickup_zone'].isin(selected_zones)]
sns.pairplot(data=df_sample, vars=['distance', 'fare', 'tip', 'total'], hue='pickup_zone')
plt.show()

#sns.pairplot(data=df1, vars=["distance", "fare", "tip", "total"], hue="pickup_zone")
#plt.show()
```

```python
# Violin Plot - Plot a violin plot to show the distribution of fare for each payment method. 
# Use the payment method as the categorical axis and fare as the numeric axis to visualize its distribution.

sns.violinplot(data=df1, x="payment", y="fare")
plt.title("Fare Distribution by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Fare")
plt.show()
```
