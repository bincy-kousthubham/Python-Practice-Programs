### Numpy Array Operations:
**Problem Statement:** To learn how to use NumPy for creating arrays, performing basic array operations, and conducting numerical computations through analyzing temperature data.
**Scenario:** You are given a list of daily average temperatures (in degrees Celsius) recorded over two weeks. Your task is to use NumPy to analyze this data.

```python
#1. Create a 1D NumPy Array: Create a 1D NumPy array named temperatures_w1 for Week 1 with the following values:[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]
import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print ("1D NumPy Array: ", temperatures_w1)
```

```python
#2. Inspection and Properties: Inspect the shape, data type, and the number of elements in the array. Inspect the shape of the temperatures_w1 array.
print("Shape: ", temperatures_w1.shape)
print("Data type: ", temperatures_w1.dtype)
print("Number of elements: ", temperatures_w1.size)
```

```python
#3. Array Operations: Convert the temperatures_w1 from Celsius to Fahrenheit using the formula: Fahrenheit = (Celsius * 9/5) + 32

temperatures_f = (temperatures_w1 * 9/5) + 32
print("Temperature in Celsius = ", temperatures_w1)
print("Temperature in Fahrenheit = ", temperatures_f)
```

```python
#Find the maximum, minimum, and mean temperatures for the week
max_temp = np.max(temperatures_w1)
min_temp = np.min(temperatures_w1)
avg_temp = np.mean(temperatures_w1)
print("Maximum Temperature of Week1 =", max_temp)
print("Minimum Temperature of Week1 =", min_temp)
print ("Average temperature of Week1 =", avg_temp.round(2))
```

```python
#4. Array Slicing and Indexing:
#Extract the temperatures for the first three days of the week.
first_3_days = temperatures_w1[0:3]
print("First 3 days temperature = ", first_3_days)
                
#Extract the temperatures for the weekend (last two days).
last_2_days = temperatures_w1[-2:]
print("Temperatures for the weekend =", last_2_days)

#Extract the temperatures for the middle three days of the week.
middle_3_days = temperatures_w1[2:5]
print("Temperatures for the middle three days =", middle_3_days)
```

```python
#5. Create a 2D Array: - Create a 2D NumPy array named temperatures where each row represents temperatures for a different week:
# ★ Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1) . ★ Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9], 
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])
print("2D NumPy array :", temperatures)
```

#### 6. Inspect and Slice the 2D Array:
- Inspect and print the shape, data type, and total number of elements in the temperatures array.
- Extract the temperatures for each week and the weekends (last two days) for both weeks.

```python
print("Shape: ", temperatures.shape)
print("Data type: ", temperatures.dtype)
print("Number of elements: ", temperatures.size)
```

```python
#Extract Week 1 temperatures
week1 = temperatures[0]
print("Week 1 Temperature: ",week1)
#Week 1 temperatures
week2=temperatures[1]
print("Week 2 Temperatures: ",week2)
#Weekend temperatures of both weeks
print("Weekend temperature of Week 1: ", week1[-2:])
print("Weekend temperatures of Week 2: ", week2[-2:])
```

#### Pandas Series:
1. Creating Pandas Series:
    - Create a Pandas Series named marks with the following data:
➢ Marks: 95, 92, 89, 85, 80
➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

```python
import pandas as pd
marks = pd.Series([95, 92, 89, 85, 80],
        index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

```python
#2. Indexing and Slicing:
#Use the integer index position to access the mark of the 1st rank student.
print("Mark of the 1st rank student: ", marks.iloc[0])

#Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.
print("Marks of the top 3 ranks: ") 
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

#Use the iloc accessor to retrieve the mark of the 3rd rank student
print("Mark of the 3rd rank student: ", marks.iloc[2])

#Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.
print("Ranks where the marks are greater than 90 : ")
filtered_marks = marks[marks > 90]
print(filtered_marks)
```

```python
#3. Manipulating Series:
#Modify the mark of the student with the 1st rank to 100.
new_mark = marks.loc['Rank1'] =  100
print(marks)
```

```python
#Remove the entry corresponding to the last rank from the Series.
marks = marks.drop(['Rank5'])
print(marks)
```

```python
#Compute the CGPA by dividing each mark in the Series by 10.
cgpa = marks/10
print("CGPA of the marks = ")
print(cgpa)
```

#### Pandas DataFrame:
1. Creating Pandas DataFrame: Create a Pandas DataFrame named transactions with the following data:
  - TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110
  - ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
  - Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'
  - Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
transactions
```

```python
#2. Data Exploration:
#Display the transactions DataFrame and its basic information, including head, tail, shape, column names, and data types.
print(transactions)
transactions.info()
transactions.head()
transactions.tail()
print(transactions.shape)
print(transactions.columns)
print(transactions.dtypes)
```

```python
#Display only the 'ProductCategory' and 'Amount' columns.
selected_columns = transactions[['ProductCategory', 'Amount']]
print(selected_columns)
```

```python
#Retrieve the last 3 columns of the DataFrame using iloc or loc.
#Using iloc
last_3_columns = transactions.iloc[:, -3:]
print(last_3_columns)

#Using loc
last_3_columns = transactions.loc[:, ['ProductCategory', 'Region', 'Amount']]
print(last_3_columns)
```

```python
#Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.
filtered_data = transactions[(transactions['Region'] == 'North') &(transactions['Amount'] > 200)]
print(filtered_data)
```

```python
#Find the value counts for the 'ProductCategory' column.
category_counts = transactions['ProductCategory'].value_counts()
print(category_counts)
```

```python
#Find the unique values in the 'Region' column.
unique_regions = transactions['Region'].unique()
print(unique_regions)
```

```python
#Group by 'Region' and find the mean amount for each region.
grouped_data = transactions.groupby('Region')
mean_amount = grouped_data['Amount'].mean()
print(mean_amount)
```

```python
#3. Manipulating the DataFrame:
#Modify the 'Amount' for the transaction with TransactionID 102 to 165.
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print(transactions)
```

```python
#Add a new column 'Discount' by calculating 10% of the 'Amount'.
transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions)
```

```python
#Remove the row with TransactionID 109.
transactions = transactions.drop(transactions[transactions['TransactionID'] == 109].index)
print(transactions)
```

```python
#Delete the 'Discount' column from the DataFrame.
transactions = transactions.drop(columns=['Discount'])
print(transactions)
```
