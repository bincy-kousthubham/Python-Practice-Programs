```python
#35.Create a function to calculate the average of student marks.

def Average_marks(marks):
    total=sum(marks)
    average=total/len(marks)
    return average
marks=[]
for i in range(1,6):
    mark=float(input("Enter mark of subject "+str(i)+": "))
    marks.append(mark)
average = Average_marks(marks)
print("Average mark for all subjects =", round(average, 2))
```

```python
#37.Create a function using *args to calculate total sales.
def total_sales(*args):
    total=sum(args)
    return total
totalSales = total_sales(10000,20000,30000,45000,40000)
print("Total Sales = ", totalSales)
```

```python
#38.Demonstrate positional and keyword arguments.
#positional arguments - pass values to a function based on the order of its parameters

def employee_details1(name,age,department):
    print("Name: ", name)
    print("Age: ", age)
    print("Department: ", department)
employee_details1("Saya", 23, "HR")

#Keyword Arguments - pass values to a function by specifying the parameter names
def employee_details2(name, age, department):
    print("Name: ", name)
    print("Age: ", age)
    print("Department: ", department)
employee_details2(name="Aarush", age=25, department="IT")
```

```python
#39.Create a global variable and modify it using global.
#global variable - variable declared outside a function. It can be accessed from different parts of the program using global keyword.

salary=40000
def update_salary():
    global salary
    salary=75000
print("Previous salary: ", salary)
update_salary()
print("Current salary: ",salary)

Institute="Entri"
def display():
    global Institute
    Institute="Entri Software Pvt Ltd"
    print(Institute)

print(Institute)
display()
```
