```python
#1.String Concatenation:

string1 = "Hello "
string2 = input("Enter your name: ")

result = string1 + string2
print(result)

string3 = ", welcome to Python programming"
result = result + string3
print(result)
```

```python
#2. String Slicing and Indexing:
input = "Hello Bincy, welcome to Python programming"

#Print the fi rst character of the string.
print(input[0])

#Print the last character of the string.
print(input[-1])

#Print the fi rst 5 characters of the string.
print(input[:5])

#Print the last 11 characters of the string.
print(input[-11:])

#Print the string in reverse.
print(input[::-1])

#Use slicing and print the word “Python” from the existing string.
start=input.index("Python")
print(input[start:start + 6])
```

```python
#3. String Methods:
strM ="Python beginner tutorial"

#a. Convert the sentence to uppercase.
print(strM.upper())

#b. Convert the sentence to lowercase.
print(strM.lower())

#c. Use Capitalize and return the sentence to the original input form.
print(strM.capitalize())

#d. Count the total number of occurrences of character ‘t’ in the string.
print(strM.count("t"))

#e. Replace all occurrences of “Python” with “Machine Learning” in the input string strM = “Python beginner tutorial”
print(strM.replace("Python","Machine Learning"))
```

```python
#Tuples (Creation, Modification and Access) :
#Create 1st tuple with values -> (10, 20, 30), 2nd tuple with values -> (40, 50, 60):

tuple1=(10, 20, 30)
tuple2=(40, 50, 60)

#a. Concatenate the two tuples and store it in “t_combine”
t_combine= tuple1 + tuple2
print(t_combine)

#b. Repeat the elements of “t_combine” 3 times
repeat=t_combine*3
print(repeat)

#c. Access the 3rd element from “t_combine”
print(t_combine[2])

#d. Access the first three elements from “t_combine”
print(t_combine[:3])

#e. Access the last three elements from “t_combine”
print(t_combine[-3:])
```
